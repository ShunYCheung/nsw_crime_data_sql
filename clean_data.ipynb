```python
import pandas as pd
data = pd.ExcelFile('data/regional_population_data_2001_2025.xlsx', engine="openpyxl")
print(pd.read_excel(data, sheet_name=0, nrows=10))
```

```python
raw = pd.read_excel(data, sheet_name="Table 1", header=None)
print(raw)

years = raw.iloc[4, 2:].astype(int).tolist()
df = raw.iloc[6:].copy()
df.columns = ["lga_code", "lga_name"] + years

df = df[pd.to_numeric(df["lga_code"], errors= "coerce").notna()]
df["lga_code"] = df["lga_code"].astype(int).astype(str)

# from wide to long
pop = df.melt(id_vars=["lga_code", "lga_name"], var_name="year", value_name="population")
pop["year"] = pop["year"].astype(int) # set year to int
pop["population"] = pd.to_numeric(pop["population"])
pop = pop.dropna(subset=["population"])

pop.to_csv("data/abs_population.csv")
print(pop.head(), len(pop))
```

```python
import pandas as pd
data = pd.ExcelFile('data/LGA_Indexes_SEIFA_2021.xlsx', engine="openpyxl")
print(pd.read_excel(data, sheet_name=0, nrows=10))
```

```python
raw = pd.read_excel(data, sheet_name="Table 1", header=None)
print(raw)

df = raw.iloc[6:, [0, 1, 2, 3]].copy()
df.columns = ["lga_code", "lga_name", "irsd", "irsd_decile"]
df = df[pd.to_numeric(df["lga_code"], errors= "coerce").notna()]
df["lga_code"] = df["lga_code"].astype(int).astype(str)
df['irsd'] = pd.to_numeric(df["irsd"], errors="coerce")
df['irsd_decile'] = pd.to_numeric(df["irsd_decile"], errors="coerce")

df.to_csv('data/socio_ec_seifa_2021.csv', index=False)
print(df.head)
```
